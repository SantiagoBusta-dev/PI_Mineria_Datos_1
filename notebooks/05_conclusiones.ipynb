A partir del archivo original, el dataset consta de 8,160 filas y 8 columnas que registran datos clave como el identificador del usuario (user_id), edad (age), plan de suscripción (subscription_plan), tiempo de reproducción mensual (monthly_watch_time_mins), país (country), género favorito (favorite_genre), última fecha de acceso (last_login_date) y tickets de soporte (customer_support_tickets). Al realizar el primer vistazo con df.head(), se identificó que las variables categóricas contienen serias inconsistencias de escritura y problemas de formato.

El diagnóstico inicial reveló anomalías críticas: se detectaron 193 valores nulos en el tiempo de reproducción, 240 en género favorito y 320 en la fecha de acceso. Además, el resumen estadístico mostró datos imposibles, como edades negativas o de hasta 150 años (70 registros fuera de rango), tiempos de reproducción negativos o exagerados mayores a 30,000 minutos, y tickets de soporte de -1 o mayores a 50. A esto se sumaron más de 126 registros duplicados y texto mal escrito con hasta 26 variantes para los mismos países y 15 para los planes debido a problemas de mayúsculas y espacios.

Para solucionar esto, se ejecutó un pipeline de transformación estructurado. Primero, se imputaron los nulos usando la mediana (757.4 minutos) en el tiempo de reproducción y la moda en el género favorito. Segundo, se convirtieron las fechas a formato datetime, eliminando 384 registros inválidos o nulos de origen. Tercero, se removieron 131 registros completamente duplicados. Cuarto, se normalizó el texto mediante diccionarios, consolidando los planes en 3 categorías, los países en 7 y los géneros en español. Finalmente, se filtraron valores ilógicos, reteniendo solo edades de 0 a 90 años, tiempos de 0 a 5,000 minutos y de 0 a 20 tickets.

Al concluir la limpieza, el historial de transformaciones reflejó una excelente tasa de retención del 90.80% sobre el dataset original. Iniciando con 8,160 filas, el volumen bajó a 7,776 tras la remoción de fechas inválidas, luego a 7,645 al limpiar duplicados, y finalmente se consolidó en 7,409 filas completamente limpias y sin valores nulos tras aplicar los filtros lógicos. El resultado final fue exportado con éxito y quedó listo para las etapas de modelado.
