```python
# Importar las librerías estándar de la clase
import pandas as pd
import numpy as np

# Cargar el dataset 
df = pd.read_json('../data/raw/streaming_users_dirty.json')

# Primer vistazo a los datos
print(df.head())
```

Se muestra el tipo de dato de cada columna y la cantidad de valores no nulos, se observa que hay algunas variables que cuentan con valores faltantes

```python
# Mostrar la cantidad de valores nulos y su porcentaje
nulos = df.isnull().sum()

porcentaje = (nulos/len(df))*100

pd.DataFrame({
    "Cantidad": nulos,
    "Porcentaje": porcentaje.round(2)
})
```
